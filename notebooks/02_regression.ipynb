**Team:** Standard Deviants

This notebook loads the Clean Ames Housing dataset and performs regression analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Load packages for regression work 
import scipy
import statsmodels.formula.api as smf
```

```python
df = pd.read_csv("../data/AmesHousing_clean.csv")
```

```python
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
df[["OverallQual", "GrLivArea", "TotalBsmtSF", "YearBuilt", "SalePrice"]].head()
```

**Selected predictors:** `OverallQual`, `GrLivArea`, `TotalBsmtSF`, and `YearBuilt` are our independent variables for predicting `SalePrice`.

```python
### Multiple Linear Regression Model
```

```python
print("Multiple Linear Regression Model")
```

```python
model = smf.ols('SalePrice ~ GrLivArea + OverallQual + TotalBsmtSF + YearBuilt', data=df).fit()
```

```python
print(model.summary())
```

**Save Results:** create output txt file and save results. 

```python
with open("../results/regression_results.txt", "w") as f:
    f.write(model.summary().as_text())
```

```python
print()
print(f"The model explains {model.rsquared * 100:.1f}% of the variation in SalePrice.\n")

print(f"Holding other variables constant: \n")

print(f" - each additional square foot of living area "
      f"is associated with a ${model.params['GrLivArea']:.2f} increase in SalePrice.\n")

print(f" - each one-point increase in OverallQual "
      f"is associated with a ${model.params['OverallQual']:.2f} increase in SalePrice.\n")

print(f" - each additional square foot of basement area "
      f"is associated with a ${model.params['TotalBsmtSF']:.2f} increase in SalePrice.\n")

print(f" - each year increase in YearBuilt "
      f"is associated with a ${model.params['YearBuilt']:.2f} increase in SalePrice.")
print()
```

### Correlation Matrix for Selected Predictors 

```python
corr = df[["OverallQual", "GrLivArea", "TotalBsmtSF", "YearBuilt"]].corr().round(2)
corr
```

```python
correlations = [
    corr.loc["OverallQual", "GrLivArea"],
    corr.loc["OverallQual", "TotalBsmtSF"],
    corr.loc["OverallQual", "YearBuilt"],
    corr.loc["GrLivArea", "TotalBsmtSF"],
    corr.loc["GrLivArea", "YearBuilt"],
    corr.loc["TotalBsmtSF", "YearBuilt"]
]
print()
print(f"Correlations among the four predictors range from "
      f"{min(correlations):.2f} to {max(correlations):.2f}. "
      f"None of the predictors are highly correlated with each other, "
      f"indicating that multicollinearity does not appear to be an issue with the model.")
print()
```
