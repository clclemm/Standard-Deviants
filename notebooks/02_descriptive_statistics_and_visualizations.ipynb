This builds on the cleaned Ames Housing dataset to produce descriptive statistics and exploratory visualizations. The analysis focuses on selected housing characteristics and their relationship with SalePrice.

```python
# loading the needed packages

import pandas as pd
import matplotlib.pyplot as plt
```

```python
Homes = pd.read_csv("../data/AmesHousing_clean.csv")

print("Dataset dimensions:", Homes.shape)
```

## Descriptive Statistics

The following descriptive statistics summarize the selected variables used in the project analysis: OverallQual, GrLivArea, TotalBsmtSF, YearBuilt, and SalePrice.

```python
selected_variables = ['OverallQual', 'GrLivArea', 'TotalBsmtSF', 'YearBuilt', 'SalePrice']

descriptive_stats = Homes[selected_variables].describe()

descriptive_stats
```

### Interpretation

The descriptive statistics show substantial variation across the selected housing characteristics. SalePrice has a mean of approximately $\text{\$180,416}$ and a median of $\text{\$160,000}$, indicating that higher-priced homes increase the average sale price. OverallQual ranges from 1 to 10, with a median rating of 6, while GrLivArea has a median of approximately 1,442 square feet. YearBuilt ranges from 1872 to 2010, showing that the dataset contains homes from a wide range of construction years.

## Visualization 1: Year Built vs. Sale Price

This scatterplot examines the relationship between the year a home was built and its sale price.

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    Homes['YearBuilt'],
    Homes['SalePrice'],
    alpha=0.5,
    color='steelblue'
)

plt.xlabel('Year Built')
plt.ylabel('Sale Price')
plt.title('Year Built vs. Sale Price')

plt.show()
```

## Scatter Plot Interpretation

The scatterplot shows a generally positive relationship between YearBuilt and SalePrice. In general, newer homes tend to have higher sale prices than older homes, although there is considerable variation in price among homes built in he same period. The relationship appears more noticeable for homes built after approximately 1950, where sale prices tend to increase as the construction year becomes more recent. There are also several high-priced homes, particularly among newer properties, as well as some lower-priced homes that do not follow the overall trend. From this, it suggests that the age of a home is associated with its sale price. However, other housing characteristics could also contribute to differences in price.

## Visualization 2: Overall Quality vs. Sale Price

This boxplot compares the distribution of sale prices across different OverallQual ratings.

```python
plt.figure(figsize=(10, 6))

data = [
    Homes.loc[Homes['OverallQual'] == quality, 'SalePrice']
    for quality in sorted(Homes['OverallQual'].dropna().unique())
]

box = plt.boxplot(
    data,
    labels=sorted(Homes['OverallQual'].dropna().unique()),
    patch_artist=True
)

for boxplot in box['boxes']:
    boxplot.set_facecolor('darkorange')

for median in box['medians']:
    median.set_color('black')
    median.set_linewidth(2)

plt.xlabel('Overall Quality Rating')
plt.ylabel('Sale Price')
plt.title('Sale Price by Overall Quality Rating')

plt.show()
```

## Box Plot Interpretation

The boxplot highlights clear differences in SalePrice across the ten OverallQual categories. The median price rises consistently across the quality ratings, with the most substantial increases occurring among homes rated 7 and above. The size of the boxes also indicates that sale prices become more dispersed at the higher quality levels, meaning there is greater variation in prices among these homes. Numerous individual points fall outside the whiskers, representing properties with unusually high or low sale prices compared with other homes in the same quality category. The separation between the quality groups demonstrates that OverallQual provides a useful way to distinguish differences in the typical sale prices of homes in the dataset.
